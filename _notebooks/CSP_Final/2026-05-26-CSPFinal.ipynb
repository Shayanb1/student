      <p class="pg-close-body">Going into this year, I had the fundamentals down. I could write code, follow logic, and get something to run. What I did not know how to do was make something worth looking at. My early projects were just big blobs of text and raw output - functional, but nothing anyone would actually want to use.</p>
      <p class="pg-close-body">What this year taught me is that writing code is only part of the job. The harder part is thinking about the person on the other side - how they find your work, what makes them stay, and why they would care. That means design, presentation, and the ability to communicate what you built and why it matters.</p>
      <p class="pg-close-body">I am leaving this class knowing how to make things that people actually want to engage with. That shift - from just making something work to making something worth someone's time - is what I will carry forward.</p>
